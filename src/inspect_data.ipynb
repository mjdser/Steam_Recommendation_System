```python
import pandas as pd 

df_libraries = pd.read_csv('../new_package/07_users/libraries_top10.csv')  # Load the user ranked data from a CSV file

# Display the columns in the dataset 
print("Columns in the libraries_top10 dataset:")
print(df_libraries.columns)

df_games = pd.read_csv('../new_package/03_features/game_features.csv')  
print("Columns in the game_features dataset:")
print(df_games.columns)
```

```python
df_game_features = df_games.drop(columns=['inclusion', 'status', 'note', 'name'])
df_game_features.head()
```

```python
df_game_features.columns
```

```python
# Merge the features in df_game_features with the user ranked data in df_libraries based on the 'appid' column

id_cols = [f'top{i}_appid' for i in range(1, 11)]
feature_cols = [c for c in df_game_features.columns if c != 'appid']

long = df_libraries.melt(
    id_vars='steamid', value_vars=id_cols,
    var_name='rank', value_name='appid'
)
long = long.merge(df_game_features, on='appid', how='left')

wide = long.pivot(index='steamid', columns='rank', values=feature_cols)
wide.columns = [f'{rank}_{feat}' for feat, rank in wide.columns]

df_libraries_new = df_libraries.merge(wide, on='steamid', how='left')
print(df_libraries_new.columns)
```

```python
# save df_libraries_new to a CSV file
df_libraries_new.to_csv('../new_package/07_users/libraries_top10_with_features.csv', index=False)
```

```python
# count average number of non-zeros in each row of the new dataset
non_zero_counts = (df_libraries_new.iloc[:, 1:] != 0).sum(axis=1)
average_non_zeros = non_zero_counts.mean()
print(f"Average number of non-zero values in each row: {average_non_zeros}")
```
